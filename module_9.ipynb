```python
import math
def strong_number(num):
    s = 0
    digits = str(num)
    for i in digits:
        s += math.factorial(int(i))
    return s == num
n = int(input())
if strong_number(n):
    print("Strong Number")
else:
    print("Not Strong Number")
```
